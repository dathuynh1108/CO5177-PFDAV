## 2. Data exploration and duplicate removal

CIFAR-100 contains 60,000 RGB images across 100 classes; exact-pixel deduplication retains one representative per image, prioritizing test, then validation, then training.


```python
# Prefer the torchvision Toronto endpoint. If the download fails, use a mirror archive
# with the official MD5; torchvision verifies the archive and extracted data batches.
DATA_ROOT.mkdir(parents=True, exist_ok=True)


class CIFAR100ChecksumMirror(CIFAR100):
    url = "https://data.brainchip.com/dataset-mirror/cifar100/cifar-100-python.tar.gz"
    filename = "cifar-100-python.tar.gz"
    tgz_md5 = "eb9058c3a382ffc7106e4002c42a8d85"


try:
    train_raw = CIFAR100(str(DATA_ROOT), train=True, download=True)
    test_raw = CIFAR100(str(DATA_ROOT), train=False, download=True)
    dataset_download_source = "University of Toronto (torchvision official URL)"
except Exception as primary_error:
    print(
        f"Toronto download failed: {type(primary_error).__name__}; trying a CIFAR-100 mirror with MD5 verification."
    )
    try:
        CIFAR100ChecksumMirror(str(DATA_ROOT), train=True, download=True)
        CIFAR100ChecksumMirror(str(DATA_ROOT), train=False, download=True)
        train_raw = CIFAR100(str(DATA_ROOT), train=True, download=False)
        test_raw = CIFAR100(str(DATA_ROOT), train=False, download=False)
        dataset_download_source = "BrainChip CIFAR mirror; archive MD5 verified against torchvision official digest"
    except Exception as mirror_error:
        raise RuntimeError(
            f"Neither Toronto nor the CIFAR-100 mirror is available: {mirror_error}"
        ) from primary_error
print("Dataset download source:", dataset_download_source)
```

### Exact-image audit


```python
train_indices, val_indices = stratified_split(train_raw.targets, PROTOCOL["split_seed"])


def pixel_hash(image):
    return hashlib.sha256(np.ascontiguousarray(image).tobytes()).hexdigest()


train_hashes = [pixel_hash(image) for image in train_raw.data]
test_hashes = [pixel_hash(image) for image in test_raw.data]


# Prefer the official test representative; remove matching content from validation/train.
# Keep one representative of each exact RGB image within each split.
def unique_indices(indices, hashes, excluded=frozenset()):
    kept, seen = [], set(excluded)
    for index in indices:
        key = hashes[index]
        if key not in seen:
            kept.append(index)
            seen.add(key)
    return kept, seen
```

```python
clean_test_indices, test_hash_set = unique_indices(range(len(test_hashes)), test_hashes)
clean_val_indices, val_hash_set = unique_indices(
    val_indices, train_hashes, test_hash_set
)
clean_train_indices, train_hash_set = unique_indices(
    train_indices, train_hashes, val_hash_set
)
assert not (
    set(train_hashes[i] for i in clean_train_indices)
    & set(train_hashes[i] for i in clean_val_indices)
)
assert not (
    set(train_hashes[i] for i in clean_train_indices)
    & set(test_hashes[i] for i in clean_test_indices)
)
assert not (
    set(train_hashes[i] for i in clean_val_indices)
    & set(test_hashes[i] for i in clean_test_indices)
)
assert len(clean_test_indices) > 0
# Retain duplicate-image representatives with test > validation > train priority; save indices.
split_audit = {
    "policy": "SHA256 exact RGB-pixel deduplication; test representative has priority, then validation, then train; one representative per exact image content per split",
    "split_seed": PROTOCOL["split_seed"],
    "official_development_train": len(train_indices),
    "official_development_validation": len(val_indices),
    "official_test": len(test_hashes),
    "retained_train": len(clean_train_indices),
    "retained_validation": len(clean_val_indices),
    "retained_test": len(clean_test_indices),
    "removed_train": len(train_indices) - len(clean_train_indices),
    "removed_validation": len(val_indices) - len(clean_val_indices),
    "removed_test": len(test_hashes) - len(clean_test_indices),
    "train_indices": clean_train_indices,
    "validation_indices": clean_val_indices,
    "test_indices": clean_test_indices,
    "duplicates_considered_exact_pixels_only": True,
}
```

## 3. Augmentation, training and checkpoint selection

Training-only crops and horizontal flips vary position and orientation; validation and test use deterministic preprocessing.

The 96×96 input preserves the earlier protocol, but resizing native 32×32 images adds no detail and remains below the recommended 128 px.

Maximum validation accuracy selects checkpoints, with minimum cross-entropy breaking ties.


```python
from image.runtime import (
    atomic_json,
    atomic_csv,
    atomic_torch_save,
    state_fingerprint,
    environment_info,
    synchronize,
)
from image.data import development_data
from image.protocol import CONFIG
```

### Training and evaluation datasets


```python
# Construct datasets using the deduplication policy and protocol transforms.
torch.backends.cuda.matmul.allow_tf32 = False
torch.backends.cudnn.allow_tf32 = False
train_base = CIFAR100(
    str(DATA_ROOT),
    train=True,
    download=False,
    transform=transform(PROTOCOL, train=True),
)
val_base = CIFAR100(
    str(DATA_ROOT), train=True, download=False, transform=transform(PROTOCOL)
)
test_base = CIFAR100(
    str(DATA_ROOT), train=False, download=False, transform=transform(PROTOCOL)
)
train_datasets = {}
val_datasets = {}
test_datasets = {}
for seed in SEEDS:
    train_datasets[seed] = EpochSubset(train_base, clean_train_indices)
    val_datasets[seed] = Subset(val_base, clean_val_indices)
    test_datasets[seed] = Subset(test_base, clean_test_indices)
```

### Training loop and validation checkpoints


```python
train_metadata = {}
histories = {}
for experiment in EXPERIMENTS:
    for seed in SEEDS:
        run_dir = OUTPUT_ROOT / experiment["id"] / f"seed_{seed}"
        run_dir.mkdir(parents=True, exist_ok=False)
        log_path = run_dir / "notebook.log"
        history = []
        seed_everything(seed)
        model = build_model(experiment, initialize_pretrained=True).to(DEVICE)
        optimizer = torch.optim.AdamW(
            [p for p in model.parameters() if p.requires_grad],
            lr=experiment["lr"],
            weight_decay=PROTOCOL["weight_decay"],
            betas=tuple(PROTOCOL["betas"]),
            eps=PROTOCOL["eps"],
        )
        scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(
            optimizer, T_max=PROTOCOL["epochs"]
        )
        train_loader = loader(
            train_datasets[seed],
            PROTOCOL["batch_size"],
            WORKERS,
            torch.Generator().manual_seed(seed),
            True,
            DEVICE,
        )
        val_loader = loader(
            val_datasets[seed],
            PROTOCOL["batch_size"],
            WORKERS,
            torch.Generator().manual_seed(seed + 1),
            False,
            DEVICE,
        )
        best = None
        frozen_hash = state_fingerprint(frozen_state(model))
        train_seconds_total = 0.0
        for epoch in range(1, PROTOCOL["epochs"] + 1):
            train_datasets[seed].set_epoch(epoch, seed)
            if DEVICE.type == "cuda":
                torch.cuda.reset_peak_memory_stats(DEVICE)
            synchronize(DEVICE)
            t0 = time.perf_counter()
            lr = optimizer.param_groups[0]["lr"]
            train_metric = epoch_pass(
                model, train_loader, DEVICE, optimizer, PROTOCOL["gradient_clip"]
            )
            synchronize(DEVICE)
            train_seconds = time.perf_counter() - t0
            train_seconds_total += train_seconds
            val_metric = epoch_pass(model, val_loader, DEVICE)
            synchronize(DEVICE)
            epoch_seconds = time.perf_counter() - t0
            assert (
                state_fingerprint(frozen_state(model)) == frozen_hash
            ), "Frozen backbone changed during training."
            row = {
                "epoch": epoch,
                "lr": lr,
                "train_ce": train_metric["ce"],
                "train_accuracy": train_metric["accuracy"],
                "train_macro_f1": train_metric["macro_f1"],
                "val_ce": val_metric["ce"],
                "val_accuracy": val_metric["accuracy"],
                "val_macro_f1": val_metric["macro_f1"],
                "train_seconds": train_seconds,
                "epoch_seconds": epoch_seconds,
                "peak_gpu_memory_bytes": (
                    torch.cuda.max_memory_allocated(DEVICE)
                    if DEVICE.type == "cuda"
                    else 0
                ),
            }
            if better(val_metric["accuracy"], val_metric["ce"], best):
                best = {
                    "epoch": epoch,
                    "accuracy": val_metric["accuracy"],
                    "ce": val_metric["ce"],
                    "macro_f1": val_metric["macro_f1"],
                    "state_sha256": state_fingerprint(model.state_dict()),
                }
                atomic_torch_save(
                    run_dir / "best.pt",
                    {
                        "model": model.state_dict(),
                        "selection": best,
                        "seed": seed,
                        "experiment": experiment["id"],
                    },
                )
            history.append(row)
            scheduler.step()
            with log_path.open("a", encoding="utf-8") as log:
                log.write(
                    f"epoch={epoch}/{PROTOCOL['epochs']} train_acc={train_metric['accuracy']:.6f} val_acc={val_metric['accuracy']:.6f} val_ce={val_metric['ce']:.6f} seconds={epoch_seconds:.1f}\n"
                )
            atomic_csv(run_dir / "history.csv", history)
            print(
                f"{experiment['id']} seed={seed} epoch={epoch}/{PROTOCOL['epochs']} val_acc={val_metric['accuracy']:.4f}"
            )
        metadata = {
            "experiment": experiment,
            "seed": seed,
            "device": str(DEVICE),
            "environment": environment_info(DEVICE),
            "completed_epochs": len(history),
            "status": "trained",
            "best_validation": best,
            "train_seconds": train_seconds_total,
            "epoch_seconds": sum(r["epoch_seconds"] for r in history),
            "peak_gpu_memory_bytes": max(r["peak_gpu_memory_bytes"] for r in history),
            "retained_train": len(clean_train_indices),
            "retained_validation": len(clean_val_indices),
            "test_evaluated": False,
            "selection": "max validation accuracy; tie min validation CE",
            "split_policy": split_audit["policy"],
        }
        atomic_json(run_dir / "metadata.json", metadata)
        histories[(experiment["id"], seed)] = history
        train_metadata[(experiment["id"], seed)] = metadata
        del optimizer, scheduler, model, train_loader, val_loader
        if DEVICE.type == "cuda":
            torch.cuda.empty_cache()
```

```python
atomic_json(
    OUTPUT_ROOT / "training_summary.json",
    {
        "runs": [
            {
                "experiment": k[0],
                "seed": k[1],
                **{
                    f: v
                    for f, v in m.items()
                    if f
                    in [
                        "status",
                        "completed_epochs",
                        "best_validation",
                        "train_seconds",
                        "epoch_seconds",
                    ]
                },
            }
            for k, m in train_metadata.items()
        ]
    },
)
print(
    "Training complete; histories, best checkpoints, configuration and logs have been saved. Test evaluation has not started."
)
```

## 4. Held-out test evaluation


```python
# Evaluate the audited test split only after recording configurations, seeds and checkpoint-selection criteria.
test_results = {}
for experiment in EXPERIMENTS:
    for seed in SEEDS:
        run_dir = OUTPUT_ROOT / experiment["id"] / f"seed_{seed}"
        checkpoint = torch.load(
            run_dir / "best.pt", map_location="cpu", weights_only=False
        )
        model = build_model(experiment, initialize_pretrained=False).to(DEVICE)
        model.load_state_dict(checkpoint["model"])
        assert (
            state_fingerprint(model.state_dict())
            == checkpoint["selection"]["state_sha256"]
        )
        test_loader = loader(
            test_datasets[seed],
            PROTOCOL["batch_size"],
            WORKERS,
            torch.Generator().manual_seed(seed + 2),
            False,
            DEVICE,
        )
        evaluation = epoch_pass(model, test_loader, DEVICE, predictions=True)
        predictions = evaluation.pop("predictions")
        for prediction in predictions:
            prediction["official_index"] = clean_test_indices[prediction["index"]]
        for row in evaluation["per_class"]:
            row["class_name"] = class_names[row["class_id"]]
        evaluation.update(
            {
                "selected_epoch": checkpoint["selection"]["epoch"],
                "selection": checkpoint["selection"],
                "evaluation_count": len(clean_test_indices),
                "split_policy": split_audit["policy"],
            }
        )
        atomic_json(run_dir / "evaluation_test.json", evaluation)
        atomic_csv(run_dir / "predictions_test.csv", predictions)
        atomic_csv(run_dir / "per_class_test.csv", evaluation["per_class"])
        atomic_csv(
            run_dir / "confusion_test.csv",
            [
                {"true_class": i, **{str(j): n for j, n in enumerate(row)}}
                for i, row in enumerate(evaluation["confusion"])
            ],
        )
        metadata = json.loads((run_dir / "metadata.json").read_text())
        metadata["test_evaluated"] = True
        atomic_json(run_dir / "metadata.json", metadata)
        test_results[(experiment["id"], seed)] = evaluation
        print(
            f"{experiment['id']} seed={seed}: test acc={evaluation['accuracy']:.4f}, macro-F1={evaluation['macro_f1']:.4f}, n={len(clean_test_indices)}"
        )
        del model, checkpoint, test_loader
        if DEVICE.type == "cuda":
            torch.cuda.empty_cache()
```

## 5. Accuracy and computational cost

Summaries report mean ± sample SD; training time excludes validation, while peak memory includes it.


```python
summary_rows = []
for experiment in EXPERIMENTS:
    runs = [
        (
            seed,
            test_results[(experiment["id"], seed)],
            train_metadata[(experiment["id"], seed)],
        )
        for seed in SEEDS
    ]
    acc = np.asarray([r[1]["accuracy"] for r in runs], dtype=float)
    f1 = np.asarray([r[1]["macro_f1"] for r in runs], dtype=float)
    train_sec = np.asarray([r[2]["train_seconds"] for r in runs], dtype=float)
    peak_mib = np.asarray(
        [r[2]["peak_gpu_memory_bytes"] / 1024**2 for r in runs], dtype=float
    )
    summary_rows.append(
        {
            "experiment": experiment["id"],
            "accuracy_mean": float(acc.mean()),
            "accuracy_sd": float(acc.std(ddof=1)) if len(acc) > 1 else 0.0,
            "macro_f1_mean": float(f1.mean()),
            "macro_f1_sd": float(f1.std(ddof=1)) if len(f1) > 1 else 0.0,
            "train_seconds_mean": float(train_sec.mean()),
            "peak_memory_mib_mean": float(peak_mib.mean()),
        }
    )
```

```python
with (OUTPUT_ROOT / "metrics_summary.csv").open("w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=list(summary_rows[0]))
    writer.writeheader()
    writer.writerows(summary_rows)
```

## 6. Learning curves and confusion matrices

Shaded bands show sample SD; confusion matrices display rounded mean counts across seeds.


```python
fig, axes = plt.subplots(
    len(EXPERIMENTS), 2, figsize=(11, 3.0 * len(EXPERIMENTS)), squeeze=False
)
for row_idx, experiment in enumerate(EXPERIMENTS):
    for ax, metric in zip(axes[row_idx], ["accuracy", "ce"]):
        for split_name in ["train", "val"]:
            values = np.stack(
                [
                    [
                        row[f"{split_name}_{metric}"]
                        for row in histories[(experiment["id"], seed)]
                    ]
                    for seed in SEEDS
                ]
            )
            mean = values.mean(0)
            sd = values.std(0, ddof=1) if len(values) > 1 else np.zeros_like(mean)
            epoch = np.arange(1, len(mean) + 1)
            ax.plot(epoch, mean, label=split_name)
            ax.fill_between(epoch, mean - sd, mean + sd, alpha=0.15)
        ax.set(title=f"{experiment['id']} · {metric}", xlabel="Epoch", ylabel=metric)
        ax.grid(alpha=0.2)
        ax.legend()
fig.tight_layout()
show_figure(fig, "learning_curves.png")
```

### Confusion matrices


```python
for experiment_id in [
    i for i in ["resnet18_full", "swin_t_full"] if i in EXPERIMENT_IDS
]:
    confusions = [
        np.asarray(test_results[(experiment_id, s)]["confusion"]) for s in SEEDS
    ]
    cm = np.rint(np.mean(confusions, axis=0)).astype(int)
    fig, ax = plt.subplots(figsize=(8, 7))
    im = ax.imshow(cm, cmap="Blues", interpolation="nearest")
    ax.set(
        title=f"{experiment_id}: mean confusion matrix across seeds",
        xlabel="Predicted label",
        ylabel="True label",
    )
    fig.colorbar(im, ax=ax, label="Image count (rounded mean)")
    fig.tight_layout()
    show_figure(fig, f"{experiment_id}-confusion.png")
```
